```python

import os
import requests
import pandas as pd
from dotenv import load_dotenv
```

```python
# Load GitHub token from .env

load_dotenv()

GITHUB_TOKEN = os.getenv("GITHUB_TOKEN")

if not GITHUB_TOKEN:
    raise ValueError(
        "GITHUB_TOKEN not found. Check your .env file."
    )
```

```python
# Configure GitHub repository

GITHUB_OWNER = "jabedulislam83"
GITHUB_REPO = "retail-intelligence-platform"
GITHUB_BRANCH = "main"
```

```python
# Configure GitHub API authentication headers

HEADERS = {
    "Authorization": f"Bearer {GITHUB_TOKEN}",
    "Accept": "application/vnd.github+json",
    "X-GitHub-Api-Version": "2022-11-28"
}
```

```python
# Define GitHub CSV file paths

FILE_PATHS = {
    "sales": "data/sales.csv",
    "customers": "data/customers.csv",
    "products": "data/products.csv",
    "inventory": "data/inventory.csv"
}
```

```python
# Create function to load CSV files from GitHub REST API

def load_github_csv(owner, repo, file_path, branch="main"):

    api_url = (
        f"https://api.github.com/repos/"
        f"{owner}/{repo}/contents/{file_path}"
        f"?ref={branch}"
    )

    response = requests.get(
        api_url,
        headers=HEADERS,
        timeout=30
    )

    response.raise_for_status()

    data = response.json()

    if data.get("type") != "file":
        raise ValueError(
            f"GitHub path is not a file: {file_path}"
        )

    download_url = data.get("download_url")

    if not download_url:
        raise ValueError(
            f"No download URL found for: {file_path}"
        )

    raw_response = requests.get(
        download_url,
        timeout=60
    )

    raw_response.raise_for_status()

    df = pd.read_csv(
        raw_response.url
    )

    return df
```

```python
# Extract sales dataset from GitHub

sales_df = load_github_csv(
    GITHUB_OWNER,
    GITHUB_REPO,
    FILE_PATHS["sales"],
    GITHUB_BRANCH
)

print("Sales dataset extracted successfully.")
print("Shape:", sales_df.shape) 
```
